## Variable Naming Rules

1. Variable names must be descriptive.
2. They must start with a letter or underscore.
3. They may contain letters, numbers, or underscores.
4. Variable names are **case-sensitive**.

<details>
<summary>Variable Naming Details</summary>

- Valid: `user_name`, `_age`, `score1`
- Invalid: `1score`, `user-name`
- Example: `student_name = "Sujith"`

</details>

## Operators in Python


```python
a=9;b=2;
print(a+b)
print(a-b)
print(a*b)
print(a/b) #true division
print(a//b) #floor division-quotient
print(a%b) #modulus division- remainder
print(a**b) #exponentiation- base raised to the power of exponent
```
