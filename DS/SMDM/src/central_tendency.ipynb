```python
receiver_1=[10, 15, 20, 25, 30, 35, 40, 45, 50, 55]
receiver_2=[5, 10, 15, 20, 45, 50]
```

```python
def central_tendency(data):
    mean= sum(data)/len(data)
    if len(data)%2==0:
        median = (sorted(data)[len(data)//2]+ sorted(data)[len(data)//2-1])/2
    else:
        median = sorted(data)[len(data)//2]
    mode= max(set(data), key=data.count)
    range= max(data)-min(data)
    return mean, median, mode,range
```

```python
print("Receiver 1 Central Tendency:", central_tendency(receiver_1))
print("Receiver 2 Central Tendency:", central_tendency(receiver_2))
```
